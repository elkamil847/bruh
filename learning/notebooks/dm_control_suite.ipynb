```python
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env

# 1. Bikin environment vector biar training lebih cepet
env = make_vec_env('Humanoid-v5', n_envs=4)

# 2. Bikin model AI nya
model = PPO(
    "MlpPolicy",
    env,
    verbose=1,
    tensorboard_log="./ppo_humanoid_tensorboard/"
)

# 3. TRAINING!!!
print("Mulai training 200.000 steps... ini 5-10 menit")
model.learn(total_timesteps=200000)

# 4. Simpan model
model.save("ppo_humanoid_v5")
print("SELESAI TRAINING! Model tersimpan")
```

```python
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env

# 1. Load model yg udah ada
model = PPO.load("ppo_humanoid_v5")

# 2. Bikin env lagi
env = make_vec_env('Humanoid-v5', n_envs=4)
model.set_env(env)

print("Lanjut training 500.000 steps lagi...")
print("Total nanti jadi 700.000 steps")

# 3. Lanjutin belajar
model.learn(total_timesteps=5000000)

# 4. Simpan model baru
model.save("ppo_humanoid_v5_lanjut")
print("SELESAI! Model baru tersimpan")
```

```python
import os
os.environ['MUJOCO_GL'] = 'egl'

import gymnasium as gym
from stable_baselines3 import PPO
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time

# 1. Load model
model = PPO.load("ppo_humanoid_v5_lanjut") # pake yg baru

# 2. Bikin env
env = gym.make('Humanoid-v5', render_mode='rgb_array')
obs, _ = env.reset()

print("Nonton AI langsung... tekan Ctrl+C buat stop")

# 3. Loop real-time
for i in range(1000):
    action, _ = model.predict(obs, deterministic=True)
    obs, reward, done, _, _ = env.step(action)

    # Tampilin tiap frame
    clear_output(wait=True)
    plt.figure(figsize=(6,6))
    plt.imshow(env.render())
    plt.axis('off')
    plt.title(f"Step: {i}")
    plt.show()

    time.sleep(0.00000003) # jeda 30ms biar 300fps

    if done:
        obs, _ = env.reset()

env.close()
```
